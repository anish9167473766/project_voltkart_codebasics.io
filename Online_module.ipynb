```python
n = int(input("Input a number: "))
print(n)
```

```python
if n%2 ==0:
  print(f"{n} is an even number")
else:
    print(f"{n} is an odd number")
```

```python
message = "Number is even" if n%2 ==0 else "Number is odd"
print(message)
```

```python
if n > 10 or n%2 ==0:
  print("Yay")
else:
    print("Nay")
```

```python
monthly_sales = [42,38,33,38,40,45]

threshold = 35

for sale_amount in monthly_sales:
  if sale_amount < threshold:
    print(f"Sale {sale_amount} is BELOW threshold {threshold}")
    break
  else:
    print(f"Sale {sale_amount} is ABOVE threshold {threshold}")
```
